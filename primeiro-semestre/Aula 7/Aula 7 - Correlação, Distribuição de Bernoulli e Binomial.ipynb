# Coeficiente de Correlação de Pearson


O coeficiente de correlação de Pearson, também conhecido como correlação linear de Pearson, é uma medida estatística que **quantifica a força e a direção da relação linear entre duas variáveis contínuas**. Desenvolvido por Karl Pearson no final do século XIX, ele se tornou uma ferramenta fundamental em diversas áreas, como estatística, econometria e ciências sociais.  Sua aplicação se estende a cenários que buscam analisar a associação entre variáveis, como identificar a influência da renda familiar na educação dos filhos ou a relação entre a pressão arterial e a idade em um grupo de pacientes. Historicamente, a correlação de Pearson teve um papel crucial no desenvolvimento da estatística inferencial e na análise de dados, sendo um marco importante na compreensão das relações entre variáveis em diferentes campos do conhecimento.

A distribuição de Bernoulli é um conceito fundamental em probabilidade e estatística, servindo como base para entender outras distribuições mais complexas, como a binomial. Ela descreve a probabilidade de um evento binário, ou seja, um evento que pode ter apenas dois resultados possíveis: sucesso ou fracasso.

**Conceito:**

Imagine um experimento com apenas duas possibilidades de resultado, como o lançamento de uma moeda (cara ou coroa) ou a resposta a uma pergunta (sim ou não). A distribuição de Bernoulli modela a probabilidade de obter um desses resultados.

**Fórmula:**

A probabilidade de sucesso (X=1) é representada por *p*, enquanto a probabilidade de fracasso (X=0) é representada por *1-p*. A função de probabilidade da distribuição de Bernoulli pode ser expressa como:

P(X=x) = p^x * (1-p)^(1-x)

Onde:

* X é uma variável aleatória que representa o resultado do experimento (0 ou 1).
* p é a probabilidade de sucesso.
* 1-p é a probabilidade de fracasso.


**Aplicações:**

A distribuição de Bernoulli é aplicável em diversas situações, como:

* **Probabilidade de um indivíduo comprar um produto específico.**
* **Probabilidade de um paciente responder positivamente a um tratamento médico.**
* **Probabilidade de um dispositivo eletrônico funcionar corretamente.**
* **Probabilidade de um investidor obter lucro em um determinado investimento.**


**Relação com a distribuição binomial:**

A distribuição binomial é uma extensão da distribuição de Bernoulli para *n* ensaios independentes, onde cada ensaio segue uma distribuição de Bernoulli com a mesma probabilidade de sucesso *p*. Em outras palavras, a distribuição binomial conta o número de sucessos em uma série de experimentos de Bernoulli.

**Em resumo:**

A distribuição de Bernoulli é uma ferramenta poderosa para modelar eventos com apenas dois resultados possíveis. Ela fornece uma base sólida para entender outras distribuições importantes, como a binomial, e possui aplicações em diversas áreas, como negócios, medicina e engenharia.


# Distribuição Binomial

A distribuição binomial é um modelo probabilístico que descreve a probabilidade de obter um número específico de sucessos em uma série de *n* ensaios independentes, onde cada ensaio tem apenas dois resultados possíveis (sucesso ou fracasso) e a probabilidade de sucesso (*p*) é constante em todos os ensaios.

image.png




**Aplicações:**

A distribuição binomial é amplamente utilizada em diversos campos, como:

* **Controle de qualidade:** determinar a probabilidade de defeitos em um lote de produtos.
* **Pesquisa:** estimar a proporção da população que possui uma determinada característica.
* **Medicina:** calcular a probabilidade de sucesso de um tratamento médico.
* **Engenharia:** analisar a probabilidade de falhas em sistemas.
* **Finanças:** modelar o retorno de investimentos.


**Em resumo:** A distribuição binomial é uma ferramenta essencial para modelar eventos com probabilidade fixa em uma série de ensaios independentes, permitindo calcular a probabilidade de um número específico de sucessos.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
df = pd.read_csv("dados-completos-Palmeiras.csv")
df_selected = df[['ano', 'jogo','home_team', 'away_team',
       'stadium', 'tournament', 'home_or_away', 'home_score', 'away_score']]
df_selected.head()
```

```python
def categorize_tournament(tournament_name):
  """Categoriza o torneio com base em palavras-chave."""
  if "Paulista" in tournament_name:
    return "Campeonato Paulista"
  elif "Libertadores" in tournament_name:
    return "Libertadores"
  elif "Brasileirão" in tournament_name:
    return "Campeonato Brasileiro"
  else:
    return "Outros"

# Aplica a função à coluna 'tournament' e cria uma nova coluna 'tournament_category'
df_selected['tournament_category'] = df_selected['tournament'].apply(categorize_tournament)

# Se você quiser substituir a coluna 'tournament' pela nova categorização, faça:
# df_selected['tournament'] = df_selected['tournament'].apply(categorize_tournament)

# Imprime as primeiras linhas para verificar
print(df_selected[['tournament', 'tournament_category']].head())
```

```python
df_selected['total_goals'] = df_selected['home_score'] + df_selected['away_score']
df_goals_summary = df_selected.groupby(['jogo', 'ano','home_or_away','tournament_category'])[['total_goals', 'home_score', 'away_score']].mean()
print(df_goals_summary)
```

```python
# Calcula a proporção de jogos com 3 gols ou mais para cada combinação de campeonato e home_or_away
proportion_df = df_goals_summary.groupby(['tournament_category', 'home_or_away'])['total_goals'].apply(lambda x: (x >= 3).mean()).reset_index(name='under_2.5_goals')

print(proportion_df)
```

```python
import plotly.express as px

# Criar o gráfico de barras interativo com Plotly Express
fig = px.bar(proportion_df, x='tournament_category', y='under_2.5_goals', color='home_or_away',
             barmode='group',
             labels={'tournament_category': 'Campeonato', 'under_2.5_goals': 'Proporção de Jogos com 3+ Gols',
                     'home_or_away': 'Mandante/Visitante'},
             title='Proporção de Jogos >2.5 Gols por Campeonato e Mandante/Visitante')

# Mostrar o gráfico
fig.show()
```

```python
import plotly.express as px
import pandas as pd
from scipy.stats import binom

def gerar_distribuicao_bernoulli(proporcao_sucesso, k):
  """
  Gera a distribuição de Bernoulli para o número de jogos com mais de 2.5 gols em k jogos.

  Args:
    proporcao_sucesso: A proporção de jogos com mais de 2.5 gols (obtida do proportions_df).
    k: O número de jogos.

  Returns:
    Um DataFrame com a tabela de distribuição e um gráfico da distribuição.
  """

  # Calcula a probabilidade de cada número de jogos com mais de 2.5 gols
  probabilidades = [binom.pmf(i, k, proporcao_sucesso) for i in range(k + 1)]

  # Cria um DataFrame com a tabela de distribuição
  tabela_distribuicao = pd.DataFrame({
      'Numero de Jogos com Mais de 2.5 Gols': list(range(k + 1)),
      'Probabilidade': probabilidades
  })

  # Cria um gráfico da distribuição
  fig = px.bar(tabela_distribuicao,
               x='Numero de Jogos com Mais de 2.5 Gols',
               y='Probabilidade',
               title=f'Distribuição de Bernoulli para {k} Jogos com Probabilidade de Sucesso {proporcao_sucesso:.2f}')
  fig.show()

  return tabela_distribuicao


# Exemplo de uso da função:
# Suponha que queremos analisar a distribuição para jogos do Campeonato Brasileiro em casa com k=10 jogos
# Verificar se a combinação existe no DataFrame antes de acessar o valor
filtered_df = proportion_df[(proportion_df['tournament_category'] == 'Libertadores') & (proportion_df['home_or_away'] == 'home')]
if not filtered_df.empty:
  proporcao_brasileiro_casa = filtered_df['under_2.5_goals'].iloc[0]
else:
  print("Combinação não encontrada no DataFrame. Verifique os valores de 'tournament_category' e 'home_or_away'.")
  proporcao_brasileiro_casa = 0  # ou outro valor padrão adequado

k_jogos = 10

tabela_distribuicao_brasileiro_casa = gerar_distribuicao_bernoulli(
    proporcao_brasileiro_casa, k_jogos)

# Você pode usar a função para outras combinações de campeonato e mandante/visitante
# acessando as proporções de sucesso do DataFrame proportion_df e definindo o k desejado.
```

# Distribuição de Poisson

image.png

A Distribuição de Poisson é um modelo probabilístico que descreve a probabilidade de um **número específico de eventos ocorrer em um intervalo de tempo** ou lugar fixo, se esses eventos ocorrem independentemente a uma taxa média constante. Em outras palavras, ela mede a probabilidade de uma determinada quantidade de eventos ocorrerem em um período de tempo, **considerando a taxa média (lambda) em que esses eventos acontecem**.

Por exemplo, a Distribuição de Poisson pode ser usada para modelar a quantidade de clientes que chegam a uma loja em uma hora, o número de carros que passam por um determinado ponto em uma rodovia em um dia, ou a quantidade de chamadas telefônicas recebidas por um call center em uma hora.

Essa distribuição é amplamente utilizada em diversas áreas, como na análise de filas de espera, na previsão da demanda de produtos, na análise de riscos e em outras áreas que envolvam eventos aleatórios que ocorrem em um determinado período de tempo ou espaço.


```python
import plotly.graph_objects as go

# Agrupar os dados para calcular a média de gols por torneio e mandante/visitante
lambda_values = df_goals_summary.groupby(['tournament_category', 'home_or_away'])['total_goals'].mean().reset_index(name='lambda')
print(lambda_values)
# Criar uma lista para armazenar os gráficos de cada combinação
fig_poisson = go.Figure()

# Iterar pelas combinações de torneio e mandante/visitante
for index, row in lambda_values.iterrows():
  tournament = row['tournament_category']
  home_or_away = row['home_or_away']
  lambda_val = row['lambda']

  # Calcular a probabilidade para cada número de gols usando a distribuição de Poisson
  x_values = np.arange(0, int(lambda_val) + 6)  # Definir o range de x de 0 até a média + 5
  probabilities = [np.exp(-lambda_val) * (lambda_val**x) / np.math.factorial(x) for x in x_values]

  # Adicionar o gráfico de linha para cada combinação
  fig_poisson.add_trace(go.Scatter(x=x_values, y=probabilities, mode='lines',
                                 name=f'{tournament} - {home_or_away}',
                                 visible=(index == 0)))

# Criar botões para alternar entre os gráficos
buttons = []
for i in range(len(lambda_values)):
  visibility = [False] * len(lambda_values)
  visibility[i] = True
  buttons.append(dict(label=f"{lambda_values['tournament_category'][i]} - {lambda_values['home_or_away'][i]}",
                      method="update",
                      args=[{"visible": visibility}]))

# Adicionar o menu de botões para alternar entre as distribuições
fig_poisson.update_layout(
    updatemenus=[dict(
        active=0,
        buttons=buttons,
        x=0.1,
        y=1.15
    )],
    title='Distribuição de Poisson para o Número Total de Gols',
    xaxis_title='Número Total de Gols',
    yaxis_title='Probabilidade'
)

fig_poisson.show()
```
